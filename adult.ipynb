### 8. (3đ) Tạo pie chart cho cột workclass và nhận xét xem nhóm nào chiếm nhiều nhất.

```python
import matplotlib.pyplot as plt

counts = df['workclass'].value_counts(normalize=True) * 100

threshold = 0.05
counts_cleaned = counts[counts >= threshold]

plt.figure(figsize=(10, 8))

plt.pie(
    counts_cleaned,
    labels=counts_cleaned.index,
    autopct='%1.1f%%',
    startangle=140,
    textprops={'fontsize': 10}
)

plt.title("Tỷ lệ phân phối Workclass\n(Tự động loại bỏ các nhóm < 0.05%)", fontsize=15, fontweight='bold', pad=20)
plt.axis('equal')

plt.show()

dropped_groups = counts[counts < threshold]
print("-" * 50)
print("BÁO CÁO LỌC DỮ LIỆU NHIỄU:")
if not dropped_groups.empty:
    print("Đã loại bỏ các nhóm sau do tỷ lệ quá nhỏ (gây ra hiện tượng 0.0%):")
    for group_name, group_pct in dropped_groups.items():
        print(f"  - {group_name}: {group_pct:.4f}%")
else:
    print("Không phát hiện nhóm nào dưới ngưỡng 0.05%.")
print("-" * 50)
```

### 21. (6đ) Số giờ tối đa một người làm việc mỗi tuần (hours-per-week feature) là bao nhiêu? Có bao nhiêu người làm việc với số giờ như vậy và tỷ lệ những người kiếm được nhiều (>50K) trong số đó là bao nhiêu?

```python
# You code here
import pandas as pd

max_hours = df['hours-per-week'].max()

hard_workers = df[df['hours-per-week'] == max_hours]
num_hard_workers = len(hard_workers)

rich_hard_workers = len(hard_workers[hard_workers['salary'] == '>50K'])

if num_hard_workers > 0:
    rich_percentage = (rich_hard_workers / num_hard_workers) * 100
else:
    rich_percentage = 0.0

print("-" * 50)
print(f"KẾT QUẢ PHÂN TÍCH:")
print(f"1. Số giờ làm việc tối đa mỗi tuần: {max_hours} giờ")
print(f"2. Số lượng người làm việc {max_hours} giờ/tuần: {num_hard_workers} người")
print(f"3. Tỷ lệ người kiếm >50K trong nhóm này: {rich_percentage:.2f}% (tương đương {rich_hard_workers}/{num_hard_workers} người)")
print("-" * 50)
```

### 22. (10đ) Tính thời gian làm việc trung bình (số giờ mỗi tuần) của những người kiếm được ít và nhiều (salary feature) cho mỗi quốc gia (native-country feature). Với đất nước Nhật Bản thì kết quả như nào?

```python
# You code here
import pandas as pd

country_salary_stats = df.groupby(['native-country', 'salary'])['hours-per-week'].mean()

stats_table = country_salary_stats.unstack()

print("BẢNG THỜI GIAN LÀM VIỆC TRUNG BÌNH THEO QUỐC GIA:")
display(stats_table)

japan_stats = stats_table.loc['Japan']

print("-" * 50)
print(f"KẾT QUẢ TẠI NHẬT BẢN (JAPAN):")
print(f"- Nhóm kiếm được ít (<=50K): {japan_stats['<=50K']:.2f} giờ/tuần")
print(f"- Nhóm kiếm được nhiều (>50K): {japan_stats['>50K']:.2f} giờ/tuần")
print("-" * 50)
```

### 23. (10đ) Số người da trắng (white in race feature) kiếm được số tiền >50k ở nước Mỹ chiếm bao nhiêu phần trăm. Câu hỏi tương tự với người da đen (black in race feature).

```python
# You code here
import pandas as pd

df_us = df[df['native-country'] == 'United-States']

def get_rich_percentage_by_race(race_name):
    race_group = df_us[df_us['race'] == race_name]

    rich_count = len(race_group[race_group['salary'] == '>50K'])

    if len(race_group) > 0:
        return (rich_count / len(race_group)) * 100
    return 0.0

white_rich_pct = get_rich_percentage_by_race('White')
black_rich_pct = get_rich_percentage_by_race('Black')

print("-" * 50)
print(f"TỶ LỆ KIẾM >50K TẠI MỸ (UNITED-STATES):")
print(f"- Người Da trắng (White): {white_rich_pct:.2f}%")
print(f"- Người Da đen (Black): {black_rich_pct:.2f}%")
print("-" * 50)
```
